```python
status = "menikah"
umur= 17
if umur >= 17 and status == "menikah":
    print("boleh nyoblos")
else:
    print("nikah dulu")

menikah = True
if umur >= 17 or status == "menikah":
    print("boleh nyoblos")
else:
    print("nikah dulu")

menikah = True
if umur >= 17 and not menikah:
    print("boleh nyoblos")
else:
    print("nikah dulu")
```

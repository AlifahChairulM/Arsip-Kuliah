Membuat Grafik

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python

x = np.array([1,2,3,4])
y = x**2

plt.title(
    'Grafik sederhana',
    fontsize =  20,
    fontstyle = 'italic',
    family = 'times new roman',
    color = 'blue',
)

plt.xlabel(
    "Sumbu X",
    color = 'pink',
    fontweight = 'bold', # ukuran ketebalan misal 200
    labelpad = 25,
)

plt.ylabel(
    "Sumbu Y",
    color = 'blue',
    fontweight = 'bold', # ukuran ketebalan misal 200
    labelpad = 25,
)

plt.grid(
    True,
    linestyle = '-.',
    linewidth = 0.3
)

plt.plot(
    x,y,
    marker = 'o',
    color = "#b3e9ff",
    linestyle = '--', #jenis garis
    linewidth = 2, #lebar garis
    markersize = 15,
    label = "$y = x^2$" #keterangan
)

# buat legenda
plt.legend(
    loc = 'upper left', #lokasi kiri atas
    fontsize = 12,
    frameon = True,
    shadow = True, #kasih bayangan
)

plt.show()
```

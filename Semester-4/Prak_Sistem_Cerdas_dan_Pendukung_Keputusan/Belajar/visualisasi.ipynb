Load dan Visualisasi Data

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data_hp.csv')
# display(df.head(15)) # muncul urut
display(df.sample(5)) # data muncul acak
```

## scatter plot

```python
plt.figure(figsize=(8,5)) #hori, verti
plt.scatter(
    df['Kamera_MP'], df['Harga_Juta'],
    color = 'red',
    marker = 'o'
)

plt.title ('Hubungan Resolusi Kamera vs Harga HP')
plt.xlabel('Resolusi Kamera (MP)')
plt.ylabel('Harga Juta (RP)')
plt.grid(linestyle = '--')
plt.show()
```

##bar plot untuk bandingkan 2 kolom numerik/ angka

```python
plt.figure(figsize=(10,5)) # ukuran  canvasnya
plt.bar(
    df['ID_HP'], df['Terjual_Bulan_Ini'], 
    color = 'skyblue'
)

plt.title('Total penjualan bulan ini')
plt.xlabel('ID Smartphone')
plt.ylabel('Jumlah Terjual (Unit)')
plt.show()
```

```python
##pie chart presentase berdasar warna
```

```python
penjualan_merek = df.groupby('Merek')['Terjual_Bulan_Ini'].sum() # ini dijumlahkan berdasar merek
#print(penjualan_merek)

plt.figure(figsize=(7,7))
plt.pie(
    penjualan_merek,
    labels = penjualan_merek.index,
    autopct = '%1.1f%%', #munculi angka belakang koma
    startangle=90, #sudut lawan arah jarum jam
    #colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
    )
plt.title('Presentase Penjualan Berdasarkan Merek')
plt.show()
```

##buat grafik bersebelahan

```python
plt.figure(figsize=(12,5))

#plt.subplot(baris, kolom, urutan)
plt.subplot(1,2,1)
plt.scatter(df['Kamera_MP'], df['Harga_Juta'], color = 'purple')
plt.title('Kamera vs Harga')
plt.xlabel('Kamera (MP)')
plt.ylabel('Harga (Juta)')
plt.grid(linestyle = '--')

plt.subplot(1,2,2)
plt.scatter(df['RAM_GB'], df['Harga_Juta'], color = 'green')
plt.title('RAM vs Harga')
plt.xlabel('RAM (GB)')
plt.ylabel('Harga (Juta)')
plt.grid(linestyle = '--')

#plt. tight_layout()
plt.show()
```

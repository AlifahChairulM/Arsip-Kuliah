##### ***Import Library***

```python
# Library 
import pandas as pd
import matplotlib.pyplot as plt

df= pd.read_csv('data_merchandise.csv')
```

##### **Soal**

1. Tampilkan data merchandise sebanyak 8 baris 

```python
# Code
df.head(8)
```

2. Tampilkan hubungan antara harga produk dan jumlah terjualnya produk dari data merchandise yang disajikan dalam scatter plot

```python
plt.figure(figsize=(10,5))
plt.scatter(df['Harga (Rp)'], 
            df['Terjual'], 
            color='skyblue', 
            marker='o')

plt.title('Hubungan Harga Produk vs Kuantitas Terjual', fontweight='bold')
plt.xlabel('Harga (Rp)')
plt.ylabel('Jumlah Terjual')
plt.show()
```

3. Tampilkan 5 Grup dengan total pendapatan tertinggi dalam bentuk grafik batang

```python
plt.figure(figsize=(6, 4))

# itung pendapatan
df['Pendapatan'] = df['Harga (Rp)'] * df['Terjual']

# ambil 5 grup dengan total pendapatan tertinggi
top_5_grup = df.groupby('Grup')['Pendapatan'].sum().sort_values(ascending=False).head(5)

top_5_grup.plot(kind='bar', 
                color=["#4AFF47", "#C184EA", "#F49EB9", "#C4F8F7", 'black'], 
                edgecolor='black',
                width=0.8) #ben gendut batang e

plt.title('Top 5 Grup Berdasarkan Pendapatan', fontweight='bold')
plt.xlabel('')
plt.ylabel('')
plt.xticks(rotation=0)

#garis atas ilang
ax= plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
```

4. Tampilkan proporsi kategori produk yang terjual dalam bentuk pie chart dengan tampilan persentase 1 angka dibelakang koma dan beri jarak pada kategori photocard

```python
plt.figure(figsize=(8, 8))

# total produk terjual berdasarkan Kategori
kategori = df.groupby('Kategori')['Terjual'].sum()

# jarak
jarak = [0.1 if cat == 'Photocard' else 0 for cat in kategori.index]

plt.pie(kategori, 
        labels=kategori.index, 
        autopct='%1.1f%%', #1 angka di belakang koma
        explode=jarak, #jaga jarah aman biar ga
        startangle=140, 
        colors=plt.cm.Pastel1.colors) #pastel yahu

# 5. Menambahkan Judul
plt.title('Proporsi Kategori Produk yang Terjual', fontsize=14, fontweight='bold')

plt.show()
```

5. Tampilkan tren penjualan harian dan sebaran lokasi pembeli dalam bentuk grafik garis dan pie chart yang ditampilkan berdampingan

```python
plt.figure(figsize=(15,5))

df['Tanggal'] = pd.to_datetime(df['Tanggal'])

# Agregasi data grafik
target_harian = df.groupby('Tanggal')['Terjual'].sum()

# Agregasi data bulet
kota = df['Kota'].value_counts()

# bulet
plt.subplot(1, 2, 1)
plt.pie(kota.values, 
        labels= kota.index, 
        autopct='%1.0f%%', 
        startangle = 0)
plt.title('Sebaran Lokasi Pembeli')

# garafik
plt.subplot(1, 2, 2)
plt.plot(target_harian.index,
         target_harian.values, 
          marker='o', 
          linestyle='--', 
          color='red', 
          linewidth=2)
plt.title('Tren Penjualan Harian')
plt.xlabel('')
plt.ylabel('')
plt.xticks() 

plt.show()
```
